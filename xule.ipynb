This sample Python code runs Arelle and the XULE plugin. Click in each cell to bring it in focus, then click the run button above to execute the commands. The cell below installs the latest version of Arelle.

```python
import os, sys
!{sys.executable} -m pip install git+https://git@github.com/arelle/arelle.git@master#egg=Arelle-release
!arelle -a
print('\nBelow is the path where Arelle was installed:\n')
%cd /srv/conda/envs/notebook/lib/python3.9/site-packages/arelle/
```

---
The cell below copies the XULE processing files into the plugin directory of the Arelle installation. **If the current directory listed on the last line of the cell above does not match the destination for XULE files on the last line of the cell below**, copy the new location above and replace what's below. __The red background does not indicate an error in this step__; it is confirmation of the process - when the * between the brackets for the step disappears, the files have finished being copied.

```python
%%bash 
set -x
svn export --force --quiet https://github.com/xbrlus/xule/trunk/plugin/xule /srv/conda/envs/notebook/lib/python3.9/site-packages/arelle/plugin/xule
```

The cell below gives information about the XULE installation, to confirm it's ready to use.

```python
!arelle --plugins 'xule' -v
```

---
The cell below contains a XULE 'file' (the file starts with 'output' on the second line; the first line is a Jupyter command for saving the file). Running the cell will save the file to the environment so it can be compiled at the next step.

If you replace the contents of the cell with the command '%load sample.xule' the contents of the file will appear below the command. Replace the cell contents with '%%writefile sample.xule' to toggle back to edit mode)

```python
%%writefile 'sample.xule'
output example
[@concept.local-name = 'Assets']
```

---
The cell below compiles the sample.xule file and runs the .zip against the instance document (report).  

```python
!arelle --plugins 'xule | transforms/SEC' -f 'https://www.sec.gov/Archives/edgar/data/1753673/000121390022027184/f20f2021_scienjoyhold.htm' --httpUserAgent "XULE-Arelle (xbrl.us; support@xbrl.us)" --xule-compile 'sample.xule' --xule-run --xule-rule-set 'sample.zip'
```
